## 2. The question: decision, action, cost of a wrong call

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Research Question

## Decision

How can website pages be grouped into meaningful content archetypes using search performance and content characteristics?

## Action

SEO specialists, digital marketers, and content managers can use these archetypes to prioritize optimization efforts, identify underperforming groups of pages, and develop strategies specific to each content type rather than evaluating thousands of pages individually.

## Cost of a Wrong Recommendation

Incorrect clustering may place pages with different behaviors into the same group, leading teams to apply ineffective optimization strategies. This wastes time, reduces resource efficiency, and may cause important content opportunities to be overlooked.

## Why Machine Learning?

Many relationships between SEO metrics are multidimensional and difficult to identify manually. Machine learning clustering algorithms can discover hidden structures within the data and organize pages into meaningful groups based on multiple features simultaneously, providing better decision support than manual analysis alone.
```

# Quick Look at the Data

The starter dataset contains **30,000 content pages** with **44 different features**, providing sufficient information for exploratory analysis and machine learning.

Several observations suggest that clustering similar pages is worthwhile:

- The dataset contains **16,262 declining pages**, showing that many pages exhibit different performance behaviors that may form distinct content archetypes.
- The **median word count is 2,877 words**, indicating considerable variation in content size across the dataset.
- The correlation between **search volume and impressions is only 0.001**, suggesting that search volume alone is a poor predictor of page visibility and that multiple features should be considered together.

These observations support the idea that content performance cannot be explained by a single metric, making clustering a suitable approach for discovering meaningful page categories.
