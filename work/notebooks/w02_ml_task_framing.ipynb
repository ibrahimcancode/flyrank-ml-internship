## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
## Target or Proxy

Since clustering is an unsupervised machine learning task, there is no predefined target label. Instead, the model will use observable features such as impressions, CTR, average position, content age, days since last update, and word count to group similar webpages. The resulting clusters act as a proxy for different content archetypes.
```
