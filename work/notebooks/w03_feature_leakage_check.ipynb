```python
import pandas as pd
import numpy as np

# Create feature dataset
features = df.copy()

# Sort by date to keep historical order
features = features.sort_values("report_date")


# Fill missing numerical values
numeric_cols = features.select_dtypes(include=np.number).columns

for col in numeric_cols:
    features[col] = features[col].fillna(0)


# Create engineered features

# Total AI traffic
ai_cols = [
    "ai_chatgpt",
    "ai_perplexity",
    "ai_gemini",
    "ai_copilot",
    "ai_claude",
    "ai_meta",
    "ai_other"
]

features["total_ai_sources"] = features[ai_cols].sum(axis=1)


# Total traffic from all channels
traffic_cols = [
    "sessions_organic",
    "sessions_direct",
    "sessions_referral",
    "sessions_social",
    "sessions_paid",
    "sessions_ai"
]

features["total_sessions"] = features[traffic_cols].sum(axis=1)


# CTR from Google Search Console
features["gsc_ctr"] = (
    features["gsc_clicks"] /
    (features["gsc_impressions"] + 1)
)


# Engagement per session
features["engagement_per_session"] = (
    features["ga4_total_engagement_sec"] /
    (features["ga4_sessions"] + 1)
)


# Convert categorical columns
categorical_cols = [
    "client_has_gsc",
    "client_has_ga4",
    "gsc_data_available",
    "ga4_data_available"
]

for col in categorical_cols:
    features[col] = features[col].fillna(False).astype(int)


features.head()
```

## Leakage Hunt

I checked the feature set for information that could reveal the answer directly.

The following leakage risks were tested:

1. Label-derived columns:
Columns containing target, label, or outcome information were searched.

2. Future information:
Columns representing future events or post-prediction measurements were checked.

3. Product or ranking signals:
Final ranking decisions or outcome-based flags were avoided.

Only pre-prediction performance signals were included.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Search feature names for possible leakage indicators

leak_keywords = [
    "target",
    "label",
    "future",
    "next",
    "outcome",
    "rank",
    "position_final"
]

possible_leaks = []

for col in features.columns:
    for keyword in leak_keywords:
        if keyword in col.lower():
            possible_leaks.append(col)

possible_leaks
```
